```python
from pathlib import Path
import sys
import pandas as pd
from scipy import stats

PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.data_loader import load_processed_data
from src.statistics import StatisticalAnalyzer

df = load_processed_data()
analyzer = StatisticalAnalyzer(df)

print("Dataset loaded successfully")
print("Rows:", len(df))
print("Columns:", len(df.columns))
```

```python
one_sample_result = analyzer.one_sample_t_test(
    "G3",
    10
)

print("ONE-SAMPLE T-TEST")
print("Null hypothesis: The average G3 grade is 10")
print("Sample mean:", round(one_sample_result["sample_mean"], 3))
print("T-statistic:", round(one_sample_result["statistic"], 3))
print("P-value:", one_sample_result["p_value"])
print("Significant:", one_sample_result["significant"])
print("Interpretation:", one_sample_result["interpretation"])
```

```python
independent_result = analyzer.independent_t_test(
    numeric_column="G3",
    group_column="sex",
    group1="F",
    group2="M"
)

print("INDEPENDENT T-TEST")
print("Null hypothesis: Female and male students have the same average G3 grade")
print("Female mean:", round(independent_result["group1_mean"], 3))
print("Male mean:", round(independent_result["group2_mean"], 3))
print("T-statistic:", round(independent_result["statistic"], 3))
print("P-value:", independent_result["p_value"])
print("Significant:", independent_result["significant"])
print("Interpretation:", independent_result["interpretation"])
```

```python
anova_result = analyzer.anova_test(
    numeric_column="G3",
    group_column="studytime"
)

print("ONE-WAY ANOVA")
print("Null hypothesis: All study-time groups have the same average G3 grade")
print("Groups:", anova_result["groups"])
print("F-statistic:", round(anova_result["statistic"], 3))
print("P-value:", anova_result["p_value"])
print("Significant:", anova_result["significant"])
print("Interpretation:", anova_result["interpretation"])
```

```python
chi_square_result = analyzer.chi_square_test(
    column1="sex",
    column2="higher"
)

print("CHI-SQUARE TEST")
print("Null hypothesis: Sex and intention to pursue higher education are independent")
print("Chi-square statistic:", round(chi_square_result["statistic"], 3))
print("P-value:", chi_square_result["p_value"])
print(
    "Degrees of freedom:",
    chi_square_result["degrees_of_freedom"]
)
print("Significant:", chi_square_result["significant"])
print("Interpretation:", chi_square_result["interpretation"])
```

```python
hypothesis_results = [
    {
        "test": "One-sample t-test",
        "variables": "G3 compared with 10",
        "statistic": one_sample_result["statistic"],
        "p_value": one_sample_result["p_value"],
        "significant": one_sample_result["significant"],
        "interpretation": one_sample_result["interpretation"]
    },
    {
        "test": "Independent t-test",
        "variables": "G3 grouped by sex",
        "statistic": independent_result["statistic"],
        "p_value": independent_result["p_value"],
        "significant": independent_result["significant"],
        "interpretation": independent_result["interpretation"]
    },
    {
        "test": "One-way ANOVA",
        "variables": "G3 grouped by studytime",
        "statistic": anova_result["statistic"],
        "p_value": anova_result["p_value"],
        "significant": anova_result["significant"],
        "interpretation": anova_result["interpretation"]
    },
    {
        "test": "Chi-square test",
        "variables": "sex and higher",
        "statistic": chi_square_result["statistic"],
        "p_value": chi_square_result["p_value"],
        "significant": chi_square_result["significant"],
        "interpretation": chi_square_result["interpretation"]
    }
]

hypothesis_summary = pd.DataFrame(hypothesis_results)

output_file = (
    PROJECT_ROOT
    / "reports"
    / "hypothesis_tests_summary.csv"
)

hypothesis_summary.to_csv(output_file, index=False)

print(hypothesis_summary)
print()
print("Saved to:", output_file)
```
