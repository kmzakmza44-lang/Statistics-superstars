# Hypothesis Testing

## Objective

Test whether final grades differ from a benchmark and across student groups, and test whether two categorical variables are associated. The significance level is $\alpha=0.05$.

## General assumptions and data note

- Observations are treated as independent students.
- `G3` is discrete and its Shapiro-Wilk result from Week 1 indicates non-normality. With 649 observations, t-tests and ANOVA are reasonably robust for mean comparisons, but results should still be interpreted cautiously.
- All tests use complete cases for the variables involved.

```python
from pathlib import Path
import sys
import pandas as pd
from scipy import stats

PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.data_loader import load_processed_data
from src.statistics import StatisticalAnalyzer

df = load_processed_data()
analyzer = StatisticalAnalyzer(df)

print("Dataset loaded successfully")
print("Rows:", len(df))
print("Columns:", len(df.columns))
```

## One-sample t-test

**Hypotheses:** $H_0: \mu_{G3}=10$; $H_1: \mu_{G3}\ne10$.

**Assumptions:** independent observations, a quantitative outcome, and an approximately normal sampling distribution of the mean. The raw grades are non-normal, but the large sample supports the sampling-distribution approximation.

```python
one_sample_result = analyzer.one_sample_t_test(
    "G3",
    10
)

print("ONE-SAMPLE T-TEST")
print("Null hypothesis: The average G3 grade is 10")
print("Sample mean:", round(one_sample_result["sample_mean"], 3))
print("T-statistic:", round(one_sample_result["statistic"], 3))
print("P-value:", one_sample_result["p_value"])
print("Significant:", one_sample_result["significant"])
print("Interpretation:", one_sample_result["interpretation"])
```

**Result and conclusion:** Mean `G3` was 11.906, $t=15.030$, $p<0.001$. Reject $H_0$; the mean final grade differs significantly from 10.

## Independent t-test

**Hypotheses:** $H_0: \mu_F=\mu_M$; $H_1: \mu_F\ne\mu_M$.

**Assumptions:** independent groups and observations, quantitative outcome, and an approximately normal sampling distribution within each group. Welch's t-test is used, so equal variances are not assumed. Grade non-normality remains a limitation, although both groups are large.

```python
independent_result = analyzer.independent_t_test(
    numeric_column="G3",
    group_column="sex",
    group1="F",
    group2="M"
)

print("INDEPENDENT T-TEST")
print("Null hypothesis: Female and male students have the same average G3 grade")
print("Female mean:", round(independent_result["group1_mean"], 3))
print("Male mean:", round(independent_result["group2_mean"], 3))
print("T-statistic:", round(independent_result["statistic"], 3))
print("P-value:", independent_result["p_value"])
print("Significant:", independent_result["significant"])
print("Interpretation:", independent_result["interpretation"])
```

**Result and conclusion:** Female mean = 12.253 and male mean = 11.406, $t=3.275$, $p=0.0011$. Reject $H_0$; the group means differ. This is an association, not evidence that sex causes grade differences.

## One-way ANOVA

**Hypotheses:** $H_0$: all four study-time groups have the same mean `G3`; $H_1$: at least one mean differs.

**Assumptions:** independent observations, approximately normal residuals within groups, and similar group variances. The discrete, non-normal grades weaken the normality assumption. The large sample gives some robustness, but a nonparametric sensitivity analysis could be considered.

```python
anova_result = analyzer.anova_test(
    numeric_column="G3",
    group_column="studytime"
)

print("ONE-WAY ANOVA")
print("Null hypothesis: All study-time groups have the same average G3 grade")
print("Groups:", anova_result["groups"])
print("F-statistic:", round(anova_result["statistic"], 3))
print("P-value:", anova_result["p_value"])
print("Significant:", anova_result["significant"])
print("Interpretation:", anova_result["interpretation"])
```

**Result:** $F=15.876$, $p<0.001$. Reject $H_0$; at least one study-time mean differs. Because the omnibus test does not identify the groups, Tukey HSD is used next while controlling family-wise error.

### Tukey HSD post-hoc comparisons

For each pair, `reject=True` indicates a significant adjusted comparison at $\alpha=0.05$.

```python
post_hoc_result = analyzer.tukey_hsd_test(
    numeric_column="G3",
    group_column="studytime"
)

post_hoc_file = PROJECT_ROOT / "reports" / "anova_tukey_hsd.csv"
post_hoc_result.to_csv(post_hoc_file, index=False)

print("TUKEY HSD POST-HOC TEST")
print(post_hoc_result.to_string(index=False))
print("Saved to:", post_hoc_file)
```

**Post-hoc conclusion:** Groups 1-2, 1-3, 1-4, and 2-3 differ significantly. Groups 2-4 and 3-4 do not. The result identifies association with study time, not causation.

## Chi-square test of independence

**Hypotheses:** $H_0$: `sex` and `higher` are independent; $H_1$: they are associated.

**Assumptions:** independent observations, mutually exclusive categories, and sufficiently large expected cell counts (normally at least 5 in each cell). The contingency table satisfies the expected-count requirement.

```python
chi_square_result = analyzer.chi_square_test(
    column1="sex",
    column2="higher"
)

print("CHI-SQUARE TEST")
print("Null hypothesis: Sex and intention to pursue higher education are independent")
print("Chi-square statistic:", round(chi_square_result["statistic"], 3))
print("P-value:", chi_square_result["p_value"])
print(
    "Degrees of freedom:",
    chi_square_result["degrees_of_freedom"]
)
print("Significant:", chi_square_result["significant"])
print("Interpretation:", chi_square_result["interpretation"])
```

**Result and conclusion:** $\chi^2(1)=1.827$, $p=0.1765$. Do not reject $H_0$; the data do not show a significant association between sex and intention to pursue higher education.

## Summary

Three mean-comparison tests were significant, while the chi-square test was not. Results must be read alongside the non-normality of grades and the observational nature of the data.
